# Module 1 · Complexity Analysis

## Concept
Before writing code we need a way to **compare algorithms independent of hardware**.
Complexity analysis measures how the running time (**time complexity**) and extra memory
(**space complexity**) grow as the input size *n* grows.

We use **asymptotic notation**:
- **Big-O `O(f)`** — upper bound (worst case). The one interviewers care about most.
- **Big-Ω `Ω(f)`** — lower bound (best case).
- **Big-Θ `Θ(f)`** — tight bound (grows exactly like f).

**Rules of thumb**
- Drop constants: `O(2n) → O(n)`.
- Drop lower-order terms: `O(n² + n) → O(n²)`.
- Nested loops over the same input multiply: two nested loops → `O(n²)`.
- Sequential loops add: `O(n) + O(n) → O(n)`.
- Halving the search space each step → `O(log n)`.

## Time & Space Complexity — reading code

| Pattern | Time | Why |
|---------|------|-----|
| Single loop `for i in range(n)` | O(n) | n iterations |
| Two nested loops | O(n²) | n × n |
| Loop that halves n each step | O(log n) | log₂n steps |
| Two separate loops | O(n) | n + n |
| Recursion with 2 calls, depth n | O(2ⁿ) | branching factor 2 |
| Building a list of size n | Space O(n) | n elements stored |
| In-place, few variables | Space O(1) | constant extra memory |

```python
import time

# O(1) - constant: one operation regardless of n
def first_element(arr):
    return arr[0] if arr else None

# O(n) - linear: touches each element once
def total(arr):
    s = 0
    for x in arr:          # n iterations
        s += x
    return s

# O(n^2) - quadratic: nested loops -> all pairs
def has_duplicate_bruteforce(arr):
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] == arr[j]:
                return True
    return False

# O(log n) - logarithmic: halve the range each step
def count_halvings(n):
    steps = 0
    while n > 1:
        n //= 2
        steps += 1
    return steps

print("first_element:", first_element([10, 20, 30]))
print("total:", total([1, 2, 3, 4]))
print("has_duplicate:", has_duplicate_bruteforce([1, 2, 3, 2]))
print("count_halvings(1024):", count_halvings(1024))  # 10, since 2^10 = 1024
```

```python
# Empirical check: watch O(n^2) grow ~4x when n doubles
import time

def timed(fn, arr):
    start = time.perf_counter()
    fn(arr)
    return time.perf_counter() - start

for n in [1000, 2000, 4000]:
    data = list(range(n))          # no duplicates -> worst case (full scan)
    t = timed(has_duplicate_bruteforce, data)
    print(f"n={n:5d}  time={t:.4f}s")
# Expect roughly 4x time each time n doubles -> confirms O(n^2)
```

# Module 2 · Arrays

## Concept
An **array** stores elements in **contiguous memory**, giving **O(1) random access** by index.
In Python the built-in `list` is a *dynamic array* (auto-resizing). Arrays are the foundation
for almost every other structure and the most common interview topic.

Core operations: traverse, insert, delete, search, reverse, rotate, and **subarray** processing.
A **subarray** is a contiguous slice; a **subsequence** keeps order but can skip elements.

Key techniques you'll reuse everywhere: **prefix sums**, **Kadane's algorithm**,
**two pointers**, and **in-place manipulation**.

## Time & Space Complexity — list (dynamic array)

| Operation | Complexity |
|-----------|-----------|
| Access `arr[i]` | O(1) |
| Update `arr[i] = x` | O(1) |
| Append at end | O(1) amortized |
| Insert/delete at front or middle | O(n) (shift elements) |
| Search (unsorted) | O(n) |
| Reverse / traverse | O(n) |
| Slice `arr[a:b]` | O(b−a) time & space |

```python
# --- Core array patterns ---

def reverse_in_place(arr):
    'Two-pointer reversal, O(n) time, O(1) space.'
    i, j = 0, len(arr) - 1
    while i < j:
        arr[i], arr[j] = arr[j], arr[i]
        i += 1
        j -= 1
    return arr

def left_rotate_by_k(arr, k):
    'Rotate array left by k using reversal trick, O(n) time, O(1) space.'
    n = len(arr)
    if n == 0:
        return arr
    k %= n
    # reverse first k, reverse rest, reverse whole
    def rev(a, i, j):
        while i < j:
            a[i], a[j] = a[j], a[i]
            i += 1; j -= 1
    rev(arr, 0, k - 1)
    rev(arr, k, n - 1)
    rev(arr, 0, n - 1)
    return arr

print("reverse:", reverse_in_place([1, 2, 3, 4, 5]))
print("rotate left by 2:", left_rotate_by_k([1, 2, 3, 4, 5, 6], 2))
```

```python
# --- Kadane's Algorithm: maximum subarray sum, O(n) ---
def max_subarray_sum(arr):
    best = cur = arr[0]
    for x in arr[1:]:
        cur = max(x, cur + x)   # extend previous subarray or start fresh
        best = max(best, cur)
    return best

print("Kadane:", max_subarray_sum([-2, 1, -3, 4, -1, 2, 1, -5, 4]))  # 6 -> [4,-1,2,1]

# --- Prefix sums: answer range-sum queries in O(1) after O(n) build ---
def build_prefix(arr):
    pre = [0] * (len(arr) + 1)
    for i, x in enumerate(arr):
        pre[i + 1] = pre[i] + x
    return pre

def range_sum(pre, l, r):        # inclusive l..r
    return pre[r + 1] - pre[l]

pre = build_prefix([3, 1, 4, 1, 5, 9])
print("sum arr[1..3]:", range_sum(pre, 1, 3))   # 1+4+1 = 6
```

```python
# --- Moore's Voting: majority element (> n/2), O(n) time O(1) space ---
def majority_element(arr):
    count, candidate = 0, None
    for x in arr:
        if count == 0:
            candidate = x
        count += 1 if x == candidate else -1
    # verify (in case no majority guaranteed)
    return candidate if arr.count(candidate) > len(arr) // 2 else None

print("majority:", majority_element([2, 2, 1, 1, 1, 2, 2]))  # 2

# --- Move zeroes to end, keep order, in place O(n) ---
def move_zeroes(arr):
    pos = 0
    for x in arr:
        if x != 0:
            arr[pos] = x
            pos += 1
    for i in range(pos, len(arr)):
        arr[i] = 0
    return arr

print("move zeroes:", move_zeroes([0, 1, 0, 3, 12]))  # [1,3,12,0,0]
```

# Module 3 · Strings

## Concept
A **string** is a sequence of characters. In Python strings are **immutable** — any
"modification" creates a new string. This immutability drives a key interview habit:
build results in a `list` and `"".join(...)` at the end (O(n)) instead of repeated `+=` (O(n²)).

Common themes: palindromes, anagrams, frequency counting, substring/subsequence checks,
character maps, and two-pointer scans.

## Time & Space Complexity — strings

| Operation | Complexity |
|-----------|-----------|
| Index / length | O(1) |
| Concatenate `a + b` | O(len a + len b) |
| Repeated `s += c` in a loop | O(n²) ⚠️ use a list instead |
| `"".join(list)` | O(n) ✅ |
| `c in s` (substring search) | O(n·m) worst |
| Slice `s[a:b]` | O(b−a) time & space |
| Sort characters | O(n log n) |

```python
# --- Palindrome check with two pointers, O(n) time O(1) space ---
def is_palindrome(s):
    i, j = 0, len(s) - 1
    while i < j:
        if s[i] != s[j]:
            return False
        i += 1; j -= 1
    return True

print("is_palindrome('racecar'):", is_palindrome("racecar"))
print("is_palindrome('hello'):", is_palindrome("hello"))

# --- Anagram check via frequency counting, O(n) ---
from collections import Counter
def are_anagrams(a, b):
    return Counter(a) == Counter(b)

print("anagrams('listen','silent'):", are_anagrams("listen", "silent"))
```

```python
# --- First non-repeating character, O(n) ---
from collections import Counter
def first_unique_char(s):
    freq = Counter(s)
    for i, c in enumerate(s):
        if freq[c] == 1:
            return i          # index of first unique char
    return -1

print("first unique in 'leetcode':", first_unique_char("leetcode"))  # 0 ('l')
print("first unique in 'aabb':", first_unique_char("aabb"))          # -1

# --- Reverse words in a sentence, O(n) ---
def reverse_words(s):
    return " ".join(s.split()[::-1])

print(reverse_words("  the sky  is blue "))  # "blue is sky the"

# --- Check subsequence, O(n) two pointers ---
def is_subsequence(sub, s):
    it = iter(s)
    return all(c in it for c in sub)

print("is_subsequence('ace','abcde'):", is_subsequence("ace", "abcde"))  # True
```

# Module 4 · Hashing

## Concept
**Hashing** maps keys to array indices via a **hash function**, giving **average O(1)**
insert, delete, and lookup. Python exposes hashing through `dict` (key→value),
`set` (unique keys), and `collections.Counter` / `defaultdict`.

Collisions (two keys → same bucket) are handled internally (open addressing in CPython).
Worst case degrades to O(n) but is extremely rare in practice.

**When to reach for a hash map:** you need to *remember what you've seen*, *count frequencies*,
*deduplicate*, or turn an O(n²) pair search into O(n).

## Time & Space Complexity — dict / set

| Operation | Average | Worst |
|-----------|---------|-------|
| Insert | O(1) | O(n) |
| Lookup / membership | O(1) | O(n) |
| Delete | O(1) | O(n) |
| Iterate all | O(n) | O(n) |

Space is **O(n)** for n stored keys. Trade memory for speed — the classic hashing bargain.

```python
# --- Two Sum: find indices of two numbers adding to target, O(n) ---
def two_sum(nums, target):
    seen = {}                     # value -> index
    for i, x in enumerate(nums):
        if target - x in seen:
            return [seen[target - x], i]
        seen[x] = i
    return []

print("two_sum:", two_sum([2, 7, 11, 15], 9))  # [0, 1]

# --- Count frequencies ---
from collections import Counter, defaultdict
def frequencies(arr):
    return dict(Counter(arr))

print("freq:", frequencies(['a', 'b', 'a', 'c', 'b', 'a']))

# --- Group anagrams (sorted signature as key) ---
def group_anagrams(words):
    groups = defaultdict(list)
    for w in words:
        key = "".join(sorted(w))
        groups[key].append(w)
    return list(groups.values())

print("anagram groups:", group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))
```

```python
# --- Longest consecutive sequence, O(n) using a set ---
def longest_consecutive(nums):
    s = set(nums)
    best = 0
    for x in s:
        if x - 1 not in s:            # only start counting at a sequence's beginning
            length = 1
            while x + length in s:
                length += 1
            best = max(best, length)
    return best

print("longest consecutive:", longest_consecutive([100, 4, 200, 1, 3, 2]))  # 4 -> [1,2,3,4]

# --- Subarray sum equals k: count subarrays, O(n) with prefix-sum hashing ---
def subarray_sum_equals_k(nums, k):
    count = 0
    prefix = 0
    seen = {0: 1}                     # prefix sum -> occurrences
    for x in nums:
        prefix += x
        count += seen.get(prefix - k, 0)
        seen[prefix] = seen.get(prefix, 0) + 1
    return count

print("subarrays sum=3:", subarray_sum_equals_k([1, 1, 1], 2))  # 2
```

# Module 5 · Linked List

## Concept
A **linked list** stores elements in **nodes**, each holding a value and a pointer to the next
node. Unlike arrays, memory is **not contiguous**, so there is **no O(1) random access** — you
must walk from the head. In return you get **O(1) insert/delete** once you hold the relevant node.

Variants: **singly** (next only), **doubly** (prev + next), **circular** (tail → head).

The signature interview techniques are the **fast & slow pointer** (Floyd's) and **in-place
reversal**. A **dummy/sentinel head** node simplifies edge cases (empty list, deleting head).

## Time & Space Complexity — singly linked list

| Operation | Complexity |
|-----------|-----------|
| Access / search by value | O(n) |
| Insert/delete at head | O(1) |
| Insert/delete at tail | O(n) (or O(1) with a tail pointer) |
| Insert/delete given the node | O(1) |
| Reverse | O(n) time, O(1) space |

```python
class ListNode:
    def __init__(self, val=0, nxt=None):
        self.val = val
        self.next = nxt

def build_list(values):
    'Helper: build a linked list from a Python list, return head.'
    dummy = ListNode()
    tail = dummy
    for v in values:
        tail.next = ListNode(v)
        tail = tail.next
    return dummy.next

def to_pylist(head):
    'Helper: convert linked list back to a Python list for printing.'
    out = []
    while head:
        out.append(head.val)
        head = head.next
    return out

head = build_list([1, 2, 3, 4, 5])
print("list:", to_pylist(head))
```

```python
# --- Reverse a singly linked list, O(n) time O(1) space ---
def reverse_list(head):
    prev = None
    cur = head
    while cur:
        nxt = cur.next      # save next
        cur.next = prev     # flip pointer
        prev = cur          # advance prev
        cur = nxt           # advance cur
    return prev             # new head

print("reversed:", to_pylist(reverse_list(build_list([1, 2, 3, 4, 5]))))

# --- Find middle node (fast & slow), O(n) ---
def middle_node(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow

print("middle of 1..5:", middle_node(build_list([1, 2, 3, 4, 5])).val)  # 3

# --- Detect a cycle (Floyd's), O(n) time O(1) space ---
def has_cycle(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            return True
    return False

print("has_cycle (linear list):", has_cycle(build_list([1, 2, 3])))  # False
```

```python
# --- Merge two sorted lists, O(n+m) ---
def merge_sorted(a, b):
    dummy = ListNode()
    tail = dummy
    while a and b:
        if a.val <= b.val:
            tail.next, a = a, a.next
        else:
            tail.next, b = b, b.next
        tail = tail.next
    tail.next = a or b       # attach the remainder
    return dummy.next

merged = merge_sorted(build_list([1, 3, 5]), build_list([2, 4, 6]))
print("merged:", to_pylist(merged))

# --- Remove N-th node from end (one pass, two pointers) ---
def remove_nth_from_end(head, n):
    dummy = ListNode(0, head)
    fast = slow = dummy
    for _ in range(n):
        fast = fast.next
    while fast.next:
        fast = fast.next
        slow = slow.next
    slow.next = slow.next.next   # skip the target
    return dummy.next

print("remove 2nd from end:", to_pylist(remove_nth_from_end(build_list([1, 2, 3, 4, 5]), 2)))
```

# Module 6 · Stack

## Concept
A **stack** is **LIFO** (Last-In-First-Out): the last element pushed is the first popped —
like a stack of plates. Operations: `push`, `pop`, `peek/top`, `is_empty`.

In Python a `list` works as a stack (`append` = push, `pop` = pop) with amortized O(1) ops.

Stacks shine for: **matching/nesting** (parentheses), **expression evaluation**,
**undo**, **DFS**, and the powerful **monotonic stack** pattern
(next greater/smaller element, histogram problems).

## Time & Space Complexity — stack

| Operation | Complexity |
|-----------|-----------|
| push / pop / peek | O(1) |
| search | O(n) |
| Space | O(n) |

Monotonic-stack algorithms process each element **at most twice** (one push, one pop) → **O(n)** total.

```python
# --- Valid parentheses (balanced brackets), O(n) ---
def is_balanced(s):
    pairs = {')': '(', ']': '[', '}': '{'}
    stack = []
    for c in s:
        if c in '([{':
            stack.append(c)
        elif c in pairs:
            if not stack or stack.pop() != pairs[c]:
                return False
    return not stack

print("'{[()]}' balanced:", is_balanced("{[()]}"))   # True
print("'([)]' balanced:", is_balanced("([)]"))        # False

# --- Evaluate postfix (Reverse Polish Notation), O(n) ---
def eval_postfix(tokens):
    stack = []
    ops = {'+', '-', '*', '/'}
    for t in tokens:
        if t in ops:
            b = stack.pop(); a = stack.pop()
            if t == '+': stack.append(a + b)
            elif t == '-': stack.append(a - b)
            elif t == '*': stack.append(a * b)
            else: stack.append(int(a / b))   # truncate toward zero
        else:
            stack.append(int(t))
    return stack[0]

print("postfix 2 3 + 4 *:", eval_postfix(["2", "3", "+", "4", "*"]))  # 20
```

```python
# --- Monotonic stack: Next Greater Element to the right, O(n) ---
def next_greater(arr):
    res = [-1] * len(arr)
    stack = []                       # holds indices; values decreasing
    for i, x in enumerate(arr):
        while stack and arr[stack[-1]] < x:
            res[stack.pop()] = x     # x is the next greater for popped index
        stack.append(i)
    return res

print("next greater:", next_greater([4, 5, 2, 25]))  # [5, 25, 25, -1]

# --- Min stack: getMin() in O(1) ---
class MinStack:
    def __init__(self):
        self.stack = []
        self.mins = []               # parallel stack of running minima
    def push(self, x):
        self.stack.append(x)
        self.mins.append(x if not self.mins else min(x, self.mins[-1]))
    def pop(self):
        self.mins.pop()
        return self.stack.pop()
    def top(self):
        return self.stack[-1]
    def get_min(self):
        return self.mins[-1]

ms = MinStack()
for v in [5, 3, 7, 2]:
    ms.push(v)
print("min:", ms.get_min())          # 2
ms.pop()
print("min after pop:", ms.get_min())# 3
```

# Module 7 · Queue

## Concept
A **queue** is **FIFO** (First-In-First-Out): the first element enqueued is the first dequeued —
like a line at a counter. Operations: `enqueue` (add rear), `dequeue` (remove front), `front`, `is_empty`.

⚠️ A Python `list` makes a **bad** queue: `pop(0)` is O(n). Use **`collections.deque`**, which gives
O(1) appends and pops at **both** ends.

Variants: **circular queue** (fixed-size ring buffer), **deque** (double-ended), and
**priority queue** (covered in the Heap module). Queues power **BFS**, scheduling, and streaming problems.

## Time & Space Complexity — deque as a queue

| Operation | list | deque |
|-----------|------|-------|
| enqueue (append right) | O(1) | O(1) |
| dequeue (pop left) | **O(n)** ⚠️ | **O(1)** ✅ |
| front / peek | O(1) | O(1) |
| Space | O(n) | O(n) |

```python
from collections import deque

# --- Queue using deque ---
q = deque()
q.append(1); q.append(2); q.append(3)     # enqueue
print("front:", q[0])                      # 1
print("dequeue:", q.popleft())             # 1
print("queue now:", list(q))               # [2, 3]

# --- Circular queue with a fixed capacity (ring buffer) ---
class CircularQueue:
    def __init__(self, k):
        self.q = [None] * k
        self.cap = k
        self.head = 0
        self.count = 0
    def enqueue(self, x):
        if self.count == self.cap:
            return False                    # full
        self.q[(self.head + self.count) % self.cap] = x
        self.count += 1
        return True
    def dequeue(self):
        if self.count == 0:
            return None                     # empty
        x = self.q[self.head]
        self.head = (self.head + 1) % self.cap
        self.count -= 1
        return x

cq = CircularQueue(3)
print([cq.enqueue(v) for v in (10, 20, 30, 40)])  # last is False (full)
print("dequeue:", cq.dequeue())                    # 10
print("enqueue 40 now:", cq.enqueue(40))           # True (space freed)
```

```python
from collections import deque

# --- Queue using two stacks (amortized O(1) dequeue) ---
class QueueViaStacks:
    def __init__(self):
        self.inbox = []
        self.outbox = []
    def enqueue(self, x):
        self.inbox.append(x)
    def dequeue(self):
        if not self.outbox:
            while self.inbox:               # move only when outbox empties
                self.outbox.append(self.inbox.pop())
        return self.outbox.pop() if self.outbox else None

qs = QueueViaStacks()
for v in (1, 2, 3):
    qs.enqueue(v)
print("dequeue order:", qs.dequeue(), qs.dequeue(), qs.dequeue())  # 1 2 3

# --- BFS on a grid: shortest steps in unweighted graph uses a queue ---
def bfs_shortest_grid(grid, start, goal):
    R, C = len(grid), len(grid[0])
    q = deque([(start, 0)])
    seen = {start}
    while q:
        (r, c), d = q.popleft()
        if (r, c) == goal:
            return d
        for dr, dc in ((1,0),(-1,0),(0,1),(0,-1)):
            nr, nc = r + dr, c + dc
            if 0 <= nr < R and 0 <= nc < C and grid[nr][nc] == 0 and (nr, nc) not in seen:
                seen.add((nr, nc))
                q.append(((nr, nc), d + 1))
    return -1

grid = [[0,0,0],
        [1,1,0],
        [0,0,0]]
print("shortest path length:", bfs_shortest_grid(grid, (0,0), (2,0)))  # 4
```

# Module 8 · Recursion

## Concept
**Recursion** is a function calling itself on a smaller input until it hits a **base case**.
Every recursive solution needs two things:
1. **Base case** — when to stop (prevents infinite recursion).
2. **Recursive case** — reduce the problem and combine results.

Recursion trades explicit loops for the **call stack**, which costs **O(depth)** memory.
It's the mental model behind **divide & conquer**, **backtracking**, **tree/graph traversal**,
and **dynamic programming**.

> Python's default recursion limit is ~1000. Deep recursion → convert to iteration or raise the limit.

## Time & Space Complexity — recursion

Use a **recurrence relation**, then solve it:
- `T(n) = T(n-1) + O(1)` → **O(n)** (linear recursion, e.g. factorial).
- `T(n) = 2·T(n-1) + O(1)` → **O(2ⁿ)** (naive Fibonacci — exponential!).
- `T(n) = 2·T(n/2) + O(n)` → **O(n log n)** (merge sort, Master theorem).

**Space** = maximum recursion depth × frame size = usually **O(depth)**.

```python
import sys
sys.setrecursionlimit(10000)

# --- Factorial: T(n)=T(n-1)+O(1) -> O(n) ---
def factorial(n):
    if n <= 1:            # base case
        return 1
    return n * factorial(n - 1)

print("5! =", factorial(5))  # 120

# --- Sum of digits ---
def sum_digits(n):
    if n == 0:
        return 0
    return n % 10 + sum_digits(n // 10)

print("sum_digits(1234):", sum_digits(1234))  # 10

# --- Naive Fibonacci: O(2^n) (shown as a WARNING pattern) ---
def fib_naive(n):
    if n < 2:
        return n
    return fib_naive(n - 1) + fib_naive(n - 2)

print("fib_naive(10):", fib_naive(10))  # 55

# --- Fibonacci with memoization: O(n) (fixes the exponential blowup) ---
from functools import lru_cache
@lru_cache(maxsize=None)
def fib_memo(n):
    if n < 2:
        return n
    return fib_memo(n - 1) + fib_memo(n - 2)

print("fib_memo(50):", fib_memo(50))  # 12586269025  (instant)
```

```python
# --- Power(x, n) in O(log n) via fast exponentiation ---
def power(x, n):
    if n == 0:
        return 1
    half = power(x, n // 2)
    return half * half if n % 2 == 0 else half * half * x

print("power(2, 10):", power(2, 10))  # 1024

# --- Generate all subsets (the backtracking skeleton) ---
def subsets(nums):
    res = []
    def backtrack(start, path):
        res.append(path[:])           # record current subset
        for i in range(start, len(nums)):
            path.append(nums[i])      # choose
            backtrack(i + 1, path)    # explore
            path.pop()                # un-choose (backtrack)
    backtrack(0, [])
    return res

print("subsets of [1,2,3]:", subsets([1, 2, 3]))

# --- Tower of Hanoi: classic recursion, 2^n - 1 moves ---
def hanoi(n, src, aux, dst, moves):
    if n == 1:
        moves.append((src, dst))
        return
    hanoi(n - 1, src, dst, aux, moves)
    moves.append((src, dst))
    hanoi(n - 1, aux, src, dst, moves)

mv = []
hanoi(3, 'A', 'B', 'C', mv)
print("hanoi 3 disks moves:", mv)
```

# Module 9 · Searching

## Concept
Two workhorses:
- **Linear search** — scan every element, works on any array, **O(n)**.
- **Binary search** — repeatedly halve a **sorted** range, **O(log n)**.

Binary search is far more than "find x in a sorted array." Its real power is
**"binary search on the answer"**: when the answer lies in a monotonic range and you can
*check feasibility* in O(n), you binary-search the answer space
(e.g. Koko eating bananas, book allocation, ship capacity).

Also master **lower_bound / upper_bound** (first index ≥ / > target) — they unlock
counting occurrences and range queries.

## Time & Space Complexity — searching

| Algorithm | Time | Space | Requires sorted? |
|-----------|------|-------|------------------|
| Linear search | O(n) | O(1) | No |
| Binary search | O(log n) | O(1) iterative | **Yes** |
| lower/upper bound | O(log n) | O(1) | Yes |
| Binary search on answer | O(n log(range)) | O(1) | monotonic check |

```python
# --- Classic binary search (avoid overflow-safe midpoint) ---
def binary_search(arr, target):
    lo, hi = 0, len(arr) - 1
    while lo <= hi:
        mid = lo + (hi - lo) // 2       # safe midpoint
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

print("index of 7:", binary_search([1, 3, 5, 7, 9, 11], 7))  # 3

# --- lower_bound: first index with arr[i] >= target ---
def lower_bound(arr, target):
    lo, hi = 0, len(arr)                # note: hi = n (exclusive)
    while lo < hi:
        mid = (lo + hi) // 2
        if arr[mid] < target:
            lo = mid + 1
        else:
            hi = mid
    return lo

# --- count occurrences using lower/upper bound ---
def count_occurrences(arr, target):
    return lower_bound(arr, target + 1) - lower_bound(arr, target)

a = [1, 2, 2, 2, 3, 4]
print("lower_bound(2):", lower_bound(a, 2))          # 1
print("count of 2:", count_occurrences(a, 2))        # 3
```

```python
# --- Search in a rotated sorted array, O(log n) ---
def search_rotated(arr, target):
    lo, hi = 0, len(arr) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if arr[mid] == target:
            return mid
        if arr[lo] <= arr[mid]:                 # left half sorted
            if arr[lo] <= target < arr[mid]:
                hi = mid - 1
            else:
                lo = mid + 1
        else:                                   # right half sorted
            if arr[mid] < target <= arr[hi]:
                lo = mid + 1
            else:
                hi = mid - 1
    return -1

print("search rotated:", search_rotated([4, 5, 6, 7, 0, 1, 2], 0))  # 4

# --- Binary search on the answer: Koko eating bananas ---
import math
def min_eating_speed(piles, h):
    def hours_needed(speed):
        return sum(math.ceil(p / speed) for p in piles)
    lo, hi = 1, max(piles)
    while lo < hi:
        mid = (lo + hi) // 2
        if hours_needed(mid) <= h:              # feasible -> try slower
            hi = mid
        else:
            lo = mid + 1
    return lo

print("min speed:", min_eating_speed([3, 6, 7, 11], 8))  # 4
```

# Module 10 · Sorting

## Concept
Sorting arranges elements in order and is a **preprocessing superpower**: many hard problems
become easy once sorted (two pointers, greedy, dedup, interval merging).

Know these by heart:
- **Bubble / Selection / Insertion** — O(n²), simple, good for tiny/nearly-sorted input.
- **Merge sort** — O(n log n), stable, divide & conquer, O(n) extra space.
- **Quick sort** — O(n log n) average / O(n²) worst, in-place, not stable.
- **Counting/Radix** — O(n+k), non-comparison, for bounded integer ranges.

Python's built-in `sorted` / `list.sort` use **Timsort**: O(n log n), stable, and very fast on
real-world (partially ordered) data. In interviews you usually *use* `sort()` but must be able to
*explain and implement* merge & quick sort.

## Time & Space Complexity — sorting

| Algorithm | Best | Average | Worst | Space | Stable |
|-----------|------|---------|-------|-------|--------|
| Bubble | O(n) | O(n²) | O(n²) | O(1) | ✅ |
| Selection | O(n²) | O(n²) | O(n²) | O(1) | ❌ |
| Insertion | O(n) | O(n²) | O(n²) | O(1) | ✅ |
| Merge | O(n log n) | O(n log n) | O(n log n) | O(n) | ✅ |
| Quick | O(n log n) | O(n log n) | O(n²) | O(log n) | ❌ |
| Timsort (Python) | O(n) | O(n log n) | O(n log n) | O(n) | ✅ |

```python
# --- Insertion sort: great for small / nearly-sorted arrays ---
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]        # shift right
            j -= 1
        arr[j + 1] = key
    return arr

print("insertion:", insertion_sort([5, 2, 4, 6, 1, 3]))

# --- Merge sort: stable, O(n log n) ---
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    # merge two sorted halves
    merged, i, j = [], 0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i]); i += 1
        else:
            merged.append(right[j]); j += 1
    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged

print("merge:", merge_sort([38, 27, 43, 3, 9, 82, 10]))
```

```python
# --- Quick sort (Lomuto partition), in place ---
def quick_sort(arr, lo=0, hi=None):
    if hi is None:
        hi = len(arr) - 1
    if lo < hi:
        p = partition(arr, lo, hi)
        quick_sort(arr, lo, p - 1)
        quick_sort(arr, p + 1, hi)
    return arr

def partition(arr, lo, hi):
    pivot = arr[hi]
    i = lo - 1
    for j in range(lo, hi):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[hi] = arr[hi], arr[i + 1]
    return i + 1

print("quick:", quick_sort([10, 7, 8, 9, 1, 5]))

# --- Custom sort keys: the interview-relevant use of built-in sort ---
words = ["banana", "kiwi", "apple", "fig"]
print("by length:", sorted(words, key=len))
print("by length desc, then alpha:", sorted(words, key=lambda w: (-len(w), w)))

# --- Merge overlapping intervals (sort first!) ---
def merge_intervals(intervals):
    intervals.sort(key=lambda x: x[0])
    out = [intervals[0]]
    for s, e in intervals[1:]:
        if s <= out[-1][1]:
            out[-1][1] = max(out[-1][1], e)   # overlap -> extend
        else:
            out.append([s, e])
    return out

print("merged intervals:", merge_intervals([[1,3],[2,6],[8,10],[15,18]]))
```

# Module 11 · Trees (Binary Tree & Binary Search Tree)

## Concept
A **tree** is a hierarchical structure of **nodes** with one **root** and no cycles. A
**binary tree** gives each node at most two children (`left`, `right`).

**Traversals** are the foundation:
- **DFS** — *inorder* (L, node, R), *preorder* (node, L, R), *postorder* (L, R, node).
- **BFS** — *level order* (queue, level by level).

A **Binary Search Tree (BST)** adds an ordering invariant: for every node,
`left subtree < node < right subtree`. This gives **O(h)** search/insert/delete where
`h` is height (O(log n) if balanced, O(n) if degenerate). An **inorder traversal of a BST
yields sorted order** — a fact interviewers love to test.

## Time & Space Complexity — trees

| Operation | Binary Tree | BST (balanced) | BST (worst / skewed) |
|-----------|-------------|----------------|----------------------|
| Traverse all | O(n) | O(n) | O(n) |
| Search | O(n) | O(log n) | O(n) |
| Insert | O(n) | O(log n) | O(n) |
| Delete | O(n) | O(log n) | O(n) |

Recursive traversal uses **O(h)** stack space (h = height). BFS uses **O(width)** queue space.

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# Build a sample tree:
#         1
#        / \
#       2   3
#      / \
#     4   5
root = TreeNode(1,
                TreeNode(2, TreeNode(4), TreeNode(5)),
                TreeNode(3))

# --- DFS traversals (recursive) ---
def inorder(node, out):
    if node:
        inorder(node.left, out); out.append(node.val); inorder(node.right, out)
    return out

def preorder(node, out):
    if node:
        out.append(node.val); preorder(node.left, out); preorder(node.right, out)
    return out

def postorder(node, out):
    if node:
        postorder(node.left, out); postorder(node.right, out); out.append(node.val)
    return out

print("inorder:  ", inorder(root, []))    # [4, 2, 5, 1, 3]
print("preorder: ", preorder(root, []))   # [1, 2, 4, 5, 3]
print("postorder:", postorder(root, []))  # [4, 5, 2, 3, 1]
```

```python
from collections import deque

# --- BFS / level-order traversal ---
def level_order(root):
    if not root:
        return []
    result, q = [], deque([root])
    while q:
        level = []
        for _ in range(len(q)):        # process one full level
            node = q.popleft()
            level.append(node.val)
            if node.left:  q.append(node.left)
            if node.right: q.append(node.right)
        result.append(level)
    return result

print("levels:", level_order(root))    # [[1], [2, 3], [4, 5]]

# --- Height / max depth ---
def height(node):
    if not node:
        return 0
    return 1 + max(height(node.left), height(node.right))

print("height:", height(root))         # 3

# --- Check if balanced (heights differ by <= 1 everywhere) ---
def is_balanced(node):
    def check(n):                      # returns height, or -1 if unbalanced
        if not n:
            return 0
        lh = check(n.left)
        if lh == -1: return -1
        rh = check(n.right)
        if rh == -1: return -1
        if abs(lh - rh) > 1: return -1
        return 1 + max(lh, rh)
    return check(node) != -1

print("balanced:", is_balanced(root))  # True
```

```python
# --- Binary Search Tree: insert, search, and validate ---
def bst_insert(root, val):
    if root is None:
        return TreeNode(val)
    if val < root.val:
        root.left = bst_insert(root.left, val)
    else:
        root.right = bst_insert(root.right, val)
    return root

def bst_search(root, val):
    while root:
        if val == root.val: return True
        root = root.left if val < root.val else root.right
    return False

def is_valid_bst(root, low=float('-inf'), high=float('inf')):
    if not root:
        return True
    if not (low < root.val < high):
        return False
    return (is_valid_bst(root.left, low, root.val) and
            is_valid_bst(root.right, root.val, high))

bst = None
for v in [5, 3, 7, 2, 4, 6, 8]:
    bst = bst_insert(bst, v)
print("inorder of BST (sorted):", inorder(bst, []))   # [2,3,4,5,6,7,8]
print("search 4:", bst_search(bst, 4))                 # True
print("search 9:", bst_search(bst, 9))                 # False
print("is valid BST:", is_valid_bst(bst))              # True

# --- Lowest Common Ancestor in a BST, O(h) ---
def lca_bst(root, p, q):
    while root:
        if p < root.val and q < root.val:
            root = root.left
        elif p > root.val and q > root.val:
            root = root.right
        else:
            return root.val
    return None

print("LCA(2,4):", lca_bst(bst, 2, 4))   # 3
```

# Module 12 · Heap / Priority Queue

## Concept
A **heap** is a complete binary tree stored in an array that maintains the **heap property**:
in a **min-heap**, every parent ≤ its children (root = minimum); in a **max-heap**, every
parent ≥ children (root = maximum). It gives **O(1) peek** at the extreme and **O(log n)**
insert/extract.

A heap *is* a **priority queue**. Python's `heapq` is a **min-heap**. For a max-heap, push
**negated** values (or `(-priority, item)` tuples).

Killer use cases: **top-k**, **k-th largest/smallest**, **merge k sorted lists**,
**running median** (two heaps), and **Dijkstra**.

## Time & Space Complexity — binary heap

| Operation | Complexity |
|-----------|-----------|
| peek min/max | O(1) |
| push | O(log n) |
| pop (extract) | O(log n) |
| heapify a list | O(n) |
| build top-k | O(n log k) |
| Space | O(n) |

```python
import heapq

# --- heapq is a MIN-heap ---
h = []
for x in [5, 1, 8, 3, 2]:
    heapq.heappush(h, x)
print("min:", h[0])                       # 1
print("pop smallest:", heapq.heappop(h))  # 1

# heapify an existing list in O(n)
data = [9, 4, 7, 1, 3]
heapq.heapify(data)
print("after heapify, min:", data[0])     # 1

# --- MAX-heap via negation ---
maxh = []
for x in [5, 1, 8, 3]:
    heapq.heappush(maxh, -x)
print("max:", -maxh[0])                    # 8

# --- k-th largest element, O(n log k) ---
def kth_largest(nums, k):
    heap = nums[:k]
    heapq.heapify(heap)                     # min-heap of size k
    for x in nums[k:]:
        if x > heap[0]:
            heapq.heapreplace(heap, x)      # pop min, push x
    return heap[0]

print("3rd largest:", kth_largest([3, 2, 1, 5, 6, 4], 3))  # 4
```

```python
import heapq

# --- Merge k sorted lists using a heap, O(N log k) ---
def merge_k_sorted(lists):
    heap = []
    for i, lst in enumerate(lists):
        if lst:
            heapq.heappush(heap, (lst[0], i, 0))   # (value, list index, elem index)
    out = []
    while heap:
        val, li, ei = heapq.heappop(heap)
        out.append(val)
        if ei + 1 < len(lists[li]):
            heapq.heappush(heap, (lists[li][ei + 1], li, ei + 1))
    return out

print("merged:", merge_k_sorted([[1, 4, 5], [1, 3, 4], [2, 6]]))

# --- Running median with two heaps ---
class MedianFinder:
    def __init__(self):
        self.lo = []   # max-heap (negated): smaller half
        self.hi = []   # min-heap: larger half
    def add(self, x):
        heapq.heappush(self.lo, -x)
        heapq.heappush(self.hi, -heapq.heappop(self.lo))   # balance values
        if len(self.hi) > len(self.lo):                    # balance sizes
            heapq.heappush(self.lo, -heapq.heappop(self.hi))
    def median(self):
        if len(self.lo) > len(self.hi):
            return -self.lo[0]
        return (-self.lo[0] + self.hi[0]) / 2

mf = MedianFinder()
for x in [5, 15, 1, 3]:
    mf.add(x)
print("running median:", mf.median())      # 4.0  (median of 1,3,5,15)
```

# Module 13 · Graphs

## Concept
A **graph** is a set of **vertices** connected by **edges**. Graphs can be **directed** or
**undirected**, **weighted** or **unweighted**, and may contain **cycles**. Grids, networks,
dependencies, and maps are all graphs.

**Representations**
- **Adjacency list** (`dict[node] -> list of neighbors`) — space O(V+E), best for sparse graphs. *Default choice.*
- **Adjacency matrix** (`V×V`) — O(V²) space, O(1) edge lookup, good for dense graphs.

**Core algorithms (medium level)**
- **BFS** — shortest path in *unweighted* graphs; level-by-level (queue).
- **DFS** — connectivity, cycle detection, components (recursion or stack).
- **Topological sort** — ordering of a DAG (course scheduling).
- **Dijkstra** — shortest path with *non-negative weights* (min-heap).

> Advanced graph algorithms (Bellman-Ford, Floyd-Warshall, MST/Union-Find, bridges/SCC)
> are deferred to a later version — master BFS/DFS/Topo/Dijkstra first.

## Time & Space Complexity — graph algorithms

Let V = vertices, E = edges.

| Algorithm | Time | Space |
|-----------|------|-------|
| BFS | O(V + E) | O(V) |
| DFS | O(V + E) | O(V) |
| Topological sort (Kahn) | O(V + E) | O(V) |
| Dijkstra (binary heap) | O((V + E) log V) | O(V) |
| Adjacency-list storage | — | O(V + E) |

```python
from collections import deque, defaultdict

def build_graph(edges, directed=False):
    g = defaultdict(list)
    for u, v in edges:
        g[u].append(v)
        if not directed:
            g[v].append(u)
    return g

# Undirected sample graph:
#   0 - 1 - 2
#   |       |
#   3 ----- 4
edges = [(0,1),(1,2),(0,3),(2,4),(3,4)]
g = build_graph(edges)

# --- BFS: visit order + shortest unweighted distances from a source ---
def bfs(g, start):
    order, dist = [], {start: 0}
    q = deque([start])
    seen = {start}
    while q:
        node = q.popleft()
        order.append(node)
        for nb in g[node]:
            if nb not in seen:
                seen.add(nb)
                dist[nb] = dist[node] + 1
                q.append(nb)
    return order, dist

order, dist = bfs(g, 0)
print("BFS order:", order)
print("distances from 0:", dict(sorted(dist.items())))
```

```python
from collections import defaultdict

# --- DFS (recursive) + connected components ---
def dfs(g, start, seen=None):
    if seen is None:
        seen = set()
    seen.add(start)
    order = [start]
    for nb in g[start]:
        if nb not in seen:
            order += dfs(g, nb, seen)
    return order

print("DFS order from 0:", dfs(g, 0))

def count_components(n, edges):
    graph = build_graph(edges)
    seen = set()
    components = 0
    for node in range(n):
        if node not in seen:
            components += 1
            # iterative DFS to mark the whole component
            stack = [node]
            while stack:
                cur = stack.pop()
                if cur in seen:
                    continue
                seen.add(cur)
                stack.extend(graph[cur])
    return components

print("connected components:", count_components(6, [(0,1),(1,2),(3,4)]))  # 3  (0-1-2, 3-4, 5)
```

```python
from collections import deque, defaultdict

# --- Cycle detection in a DIRECTED graph (DFS colors) ---
def has_cycle_directed(n, edges):
    g = defaultdict(list)
    for u, v in edges:
        g[u].append(v)
    color = [0] * n            # 0=unvisited, 1=in-stack, 2=done
    def dfs(u):
        color[u] = 1
        for v in g[u]:
            if color[v] == 1:          # back edge -> cycle
                return True
            if color[v] == 0 and dfs(v):
                return True
        color[u] = 2
        return False
    return any(color[i] == 0 and dfs(i) for i in range(n))

print("has cycle (0->1->2->0):", has_cycle_directed(3, [(0,1),(1,2),(2,0)]))  # True
print("has cycle (0->1->2):", has_cycle_directed(3, [(0,1),(1,2)]))            # False

# --- Topological sort (Kahn's algorithm, BFS on in-degrees) ---
def topological_sort(n, edges):
    g = defaultdict(list)
    indeg = [0] * n
    for u, v in edges:
        g[u].append(v)
        indeg[v] += 1
    q = deque(i for i in range(n) if indeg[i] == 0)
    order = []
    while q:
        u = q.popleft()
        order.append(u)
        for v in g[u]:
            indeg[v] -= 1
            if indeg[v] == 0:
                q.append(v)
    return order if len(order) == n else []   # empty => cycle exists

# course prerequisites: 0->1->3, 0->2->3
print("topo order:", topological_sort(4, [(0,1),(0,2),(1,3),(2,3)]))
```

```python
import heapq
from collections import defaultdict

# --- Dijkstra: shortest paths with non-negative weights ---
def dijkstra(n, weighted_edges, src):
    g = defaultdict(list)
    for u, v, w in weighted_edges:
        g[u].append((v, w))
        g[v].append((u, w))          # undirected
    dist = [float('inf')] * n
    dist[src] = 0
    pq = [(0, src)]                  # (distance, node)
    while pq:
        d, u = heapq.heappop(pq)
        if d > dist[u]:
            continue                 # stale entry
        for v, w in g[u]:
            if d + w < dist[v]:
                dist[v] = d + w
                heapq.heappush(pq, (dist[v], v))
    return dist

edges_w = [(0,1,4),(0,2,1),(2,1,2),(1,3,1),(2,3,5)]
print("shortest distances from 0:", dijkstra(4, edges_w, 0))  # [0, 3, 1, 4]
```

# Module 14 · Sliding Window

## Concept
The **sliding window** technique maintains a contiguous window (a range `[left, right]`) over an
array/string and slides it to avoid recomputing from scratch. It turns many **O(n²)** brute-force
subarray/substring problems into **O(n)**.

Two flavors:
- **Fixed-size window** — window length is a given `k` (max sum of size k, first negative per window).
- **Variable-size window** — grow `right` to include, shrink `left` to satisfy a constraint
  (longest substring without repeats, minimum window sum ≥ target).

**Signal to use it:** the problem mentions **contiguous subarray/substring** plus an
optimization/counting objective (*longest*, *shortest*, *max/min sum*, *at most k distinct*).

## Time & Space Complexity — sliding window

| Variant | Time | Space |
|---------|------|-------|
| Fixed window | O(n) | O(1) or O(k) |
| Variable window (+ hash map) | O(n) | O(k) for the map |

Each element enters and leaves the window **at most once** → each pointer moves ≤ n times → O(n).

```python
# --- Fixed window: maximum sum of any subarray of size k, O(n) ---
def max_sum_window(arr, k):
    window = sum(arr[:k])
    best = window
    for i in range(k, len(arr)):
        window += arr[i] - arr[i - k]   # add new, drop old
        best = max(best, window)
    return best

print("max sum window k=3:", max_sum_window([2, 1, 5, 1, 3, 2], 3))  # 9 -> [5,1,3]

# --- Fixed window: first negative number in every window of size k ---
from collections import deque
def first_negative_each_window(arr, k):
    negatives = deque()                 # indices of negative numbers
    res = []
    for i, x in enumerate(arr):
        if x < 0:
            negatives.append(i)
        if negatives and negatives[0] <= i - k:
            negatives.popleft()         # drop out-of-window index
        if i >= k - 1:
            res.append(arr[negatives[0]] if negatives else 0)
    return res

print("first negatives:", first_negative_each_window([-8, 2, 3, -6, 10], 2))  # [-8, 0, -6, -6]
```

```python
# --- Variable window: longest substring without repeating characters, O(n) ---
def longest_unique_substring(s):
    last = {}                           # char -> last index seen
    left = 0
    best = 0
    for right, c in enumerate(s):
        if c in last and last[c] >= left:
            left = last[c] + 1          # jump left past the duplicate
        last[c] = right
        best = max(best, right - left + 1)
    return best

print("longest unique in 'abcabcbb':", longest_unique_substring("abcabcbb"))  # 3 ('abc')

# --- Variable window: smallest subarray with sum >= target, O(n) ---
def min_subarray_len(target, arr):
    left = 0
    total = 0
    best = float('inf')
    for right, x in enumerate(arr):
        total += x
        while total >= target:          # shrink while still valid
            best = min(best, right - left + 1)
            total -= arr[left]
            left += 1
    return 0 if best == float('inf') else best

print("min len for sum>=7:", min_subarray_len(7, [2, 3, 1, 2, 4, 3]))  # 2 -> [4,3]

# --- Variable window: longest substring with at most k distinct chars ---
from collections import defaultdict
def longest_k_distinct(s, k):
    count = defaultdict(int)
    left = 0
    best = 0
    for right, c in enumerate(s):
        count[c] += 1
        while len(count) > k:
            count[s[left]] -= 1
            if count[s[left]] == 0:
                del count[s[left]]
            left += 1
        best = max(best, right - left + 1)
    return best

print("longest with <=2 distinct in 'eceba':", longest_k_distinct("eceba", 2))  # 3 ('ece')
```

# Module 15 · Two Pointers

## Concept
The **two-pointer** technique uses two indices that move through the data to replace a nested
loop, dropping **O(n²) → O(n)** (or O(n log n) with a sort). Common configurations:
- **Opposite ends** — `left` at start, `right` at end, move inward (sorted two-sum, container with most water, palindrome).
- **Same direction / fast-slow** — both start left; one advances faster (remove duplicates, cycle detection, partitioning).
- **Two arrays** — one pointer per array (merge, intersection of sorted arrays).

Two pointers usually needs the input to be **sorted** or to have some **monotonic structure**.

## Time & Space Complexity — two pointers

| Task | Time | Space |
|------|------|-------|
| On already-sorted input | O(n) | O(1) |
| When you must sort first | O(n log n) | O(1)/O(n) |
| Merge two sorted arrays | O(n + m) | O(n + m) output |

The win is **O(1) extra space** and a single pass compared to hashing or brute force.

```python
# --- Two-sum in a SORTED array (opposite ends), O(n) O(1) ---
def two_sum_sorted(arr, target):
    i, j = 0, len(arr) - 1
    while i < j:
        s = arr[i] + arr[j]
        if s == target:
            return [i, j]
        elif s < target:
            i += 1              # need a bigger sum
        else:
            j -= 1              # need a smaller sum
    return []

print("two-sum sorted:", two_sum_sorted([1, 2, 4, 7, 11, 15], 15))  # [3, 4] -> 4+11

# --- Remove duplicates from a sorted array in place (fast/slow), O(n) ---
def remove_duplicates(arr):
    if not arr:
        return 0
    slow = 0
    for fast in range(1, len(arr)):
        if arr[fast] != arr[slow]:
            slow += 1
            arr[slow] = arr[fast]
    return slow + 1             # length of unique prefix

a = [0, 0, 1, 1, 1, 2, 2, 3]
k = remove_duplicates(a)
print("unique count:", k, "-> array:", a[:k])   # 4 -> [0,1,2,3]
```

```python
# --- 3-Sum: all unique triplets summing to 0, O(n^2) ---
def three_sum(nums):
    nums.sort()
    res = []
    n = len(nums)
    for i in range(n - 2):
        if i > 0 and nums[i] == nums[i - 1]:
            continue                        # skip duplicate anchors
        left, right = i + 1, n - 1
        while left < right:
            s = nums[i] + nums[left] + nums[right]
            if s == 0:
                res.append([nums[i], nums[left], nums[right]])
                left += 1; right -= 1
                while left < right and nums[left] == nums[left - 1]:
                    left += 1               # skip duplicates
                while left < right and nums[right] == nums[right + 1]:
                    right -= 1
            elif s < 0:
                left += 1
            else:
                right -= 1
    return res

print("3-sum:", three_sum([-1, 0, 1, 2, -1, -4]))  # [[-1,-1,2], [-1,0,1]]

# --- Container with most water, O(n) ---
def max_area(height):
    i, j = 0, len(height) - 1
    best = 0
    while i < j:
        best = max(best, (j - i) * min(height[i], height[j]))
        if height[i] < height[j]:
            i += 1                          # move the shorter wall inward
        else:
            j -= 1
    return best

print("max water:", max_area([1, 8, 6, 2, 5, 4, 8, 3, 7]))  # 49
```

# Module 16 · Greedy Algorithms

## Concept
A **greedy** algorithm builds a solution step by step, always taking the **locally optimal choice**
in the hope it leads to a **globally optimal** solution. It's fast and simple — *when it works*.

Greedy is correct only when the problem has:
1. **Greedy-choice property** — a global optimum can be reached by local optimal choices.
2. **Optimal substructure** — an optimal solution contains optimal solutions to subproblems.

The usual first step is **sorting** by the right key (finish time, ratio, deadline), then a single sweep.

⚠️ Greedy is **not always correct** — e.g. coin change with arbitrary denominations needs DP.
Always sanity-check with a counterexample or a proof (exchange argument) before committing.

## Time & Space Complexity — greedy

Typically dominated by the sort:

| Step | Complexity |
|------|-----------|
| Sort input | O(n log n) |
| Single greedy sweep | O(n) |
| **Total** | **O(n log n)** |
| Space | O(1)–O(n) |

```python
# --- Activity selection: max non-overlapping activities, O(n log n) ---
def activity_selection(intervals):
    intervals.sort(key=lambda x: x[1])      # sort by finish time
    count = 0
    last_end = float('-inf')
    chosen = []
    for start, end in intervals:
        if start >= last_end:               # doesn't overlap the last chosen
            chosen.append((start, end))
            last_end = end
            count += 1
    return count, chosen

acts = [(1, 3), (2, 5), (4, 7), (1, 8), (5, 9), (8, 10)]
print("activity selection:", activity_selection(acts))

# --- Fractional knapsack: greedy by value/weight ratio, O(n log n) ---
def fractional_knapsack(items, capacity):   # items: (value, weight)
    items.sort(key=lambda it: it[0] / it[1], reverse=True)
    total = 0.0
    for value, weight in items:
        if capacity >= weight:
            total += value
            capacity -= weight
        else:
            total += value * (capacity / weight)   # take a fraction
            break
    return total

print("fractional knapsack:", fractional_knapsack([(60, 10), (100, 20), (120, 30)], 50))  # 240.0
```

```python
import heapq

# --- Minimum cost to connect n ropes (always join two smallest), O(n log n) ---
def connect_ropes(ropes):
    heapq.heapify(ropes)
    cost = 0
    while len(ropes) > 1:
        a = heapq.heappop(ropes)
        b = heapq.heappop(ropes)
        cost += a + b
        heapq.heappush(ropes, a + b)
    return cost

print("connect ropes cost:", connect_ropes([4, 3, 2, 6]))  # 29

# --- Jump game: can we reach the last index? greedy reachable frontier, O(n) ---
def can_jump(nums):
    reach = 0
    for i, jump in enumerate(nums):
        if i > reach:
            return False                    # stuck before this index
        reach = max(reach, i + jump)
    return True

print("can jump [2,3,1,1,4]:", can_jump([2, 3, 1, 1, 4]))  # True
print("can jump [3,2,1,0,4]:", can_jump([3, 2, 1, 0, 4]))  # False

# --- Assign cookies: maximize content children, O(n log n) ---
def assign_cookies(greed, sizes):
    greed.sort(); sizes.sort()
    i = j = 0
    while i < len(greed) and j < len(sizes):
        if sizes[j] >= greed[i]:            # cookie satisfies this child
            i += 1
        j += 1
    return i

print("content children:", assign_cookies([1, 2, 3], [1, 1]))  # 1
```

# Module 17 · Dynamic Programming (DP)

## Concept
**Dynamic Programming** solves problems with **overlapping subproblems** and **optimal substructure**
by solving each subproblem **once** and reusing the answer. It's *recursion + memory*.

Two styles:
- **Top-down (memoization)** — write the recursion, cache results (`@lru_cache` or a dict).
- **Bottom-up (tabulation)** — fill a table iteratively from base cases up.

**The DP recipe (use this on every DP problem):**
1. Define the **state** — what does `dp[i]` / `dp[i][j]` mean?
2. Write the **recurrence** — how does a state depend on smaller ones?
3. Set **base cases**.
4. Decide **iteration order** (so dependencies are ready).
5. **Optimize space** if only the last row/few states are needed.

Classic families: **1-D** (climbing stairs, house robber, coin change),
**2-D grid** (unique paths, min path sum), **subsequence/string** (LCS, edit distance),
**knapsack** (subset sum, 0/1 knapsack), and **LIS**.

## Time & Space Complexity — DP

Roughly: **(number of states) × (work per state)**.

| Problem | States | Time | Space |
|---------|--------|------|-------|
| Fibonacci / climbing stairs | O(n) | O(n) | O(1) (rolling) |
| House robber | O(n) | O(n) | O(1) |
| Coin change (min coins) | O(n·amount) | O(n·amount) | O(amount) |
| 0/1 Knapsack | O(n·W) | O(n·W) | O(W) (1-D) |
| LCS / edit distance | O(n·m) | O(n·m) | O(min(n,m)) |
| LIS (binary search) | O(n) | O(n log n) | O(n) |

```python
# --- Climbing stairs (Fibonacci-shaped), O(n) time O(1) space ---
def climb_stairs(n):
    a, b = 1, 1                     # ways to reach step 0 and 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

print("ways to climb 5 stairs:", climb_stairs(5))   # 8

# --- House robber: max non-adjacent sum, O(n) O(1) ---
def house_robber(nums):
    prev, cur = 0, 0
    for x in nums:
        prev, cur = cur, max(cur, prev + x)   # rob or skip current house
    return cur

print("max robbed:", house_robber([2, 7, 9, 3, 1]))  # 12  (2+9+1)

# --- Coin change: fewest coins to make amount, O(n*amount) ---
def coin_change(coins, amount):
    INF = float('inf')
    dp = [0] + [INF] * amount           # dp[a] = min coins for amount a
    for a in range(1, amount + 1):
        for c in coins:
            if c <= a:
                dp[a] = min(dp[a], dp[a - c] + 1)
    return dp[amount] if dp[amount] != INF else -1

print("min coins for 11 from [1,2,5]:", coin_change([1, 2, 5], 11))  # 3  (5+5+1)
```

```python
# --- 0/1 Knapsack (bottom-up, 1-D space optimized), O(n*W) ---
def knapsack(values, weights, W):
    dp = [0] * (W + 1)
    for i in range(len(values)):
        # iterate capacity DOWNWARD so each item is used at most once
        for w in range(W, weights[i] - 1, -1):
            dp[w] = max(dp[w], dp[w - weights[i]] + values[i])
    return dp[W]

print("knapsack:", knapsack([60, 100, 120], [10, 20, 30], 50))  # 220

# --- Longest Common Subsequence (2-D tabulation), O(n*m) ---
def lcs(a, b):
    n, m = len(a), len(b)
    dp = [[0] * (m + 1) for _ in range(n + 1)]
    for i in range(1, n + 1):
        for j in range(1, m + 1):
            if a[i - 1] == b[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])
    return dp[n][m]

print("LCS('abcde','ace'):", lcs("abcde", "ace"))   # 3

# --- Edit distance (Levenshtein), O(n*m) ---
def edit_distance(a, b):
    n, m = len(a), len(b)
    dp = [[0] * (m + 1) for _ in range(n + 1)]
    for i in range(n + 1):
        dp[i][0] = i
    for j in range(m + 1):
        dp[0][j] = j
    for i in range(1, n + 1):
        for j in range(1, m + 1):
            if a[i - 1] == b[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = 1 + min(dp[i - 1][j],      # delete
                                   dp[i][j - 1],      # insert
                                   dp[i - 1][j - 1])  # replace
    return dp[n][m]

print("edit distance('horse','ros'):", edit_distance("horse", "ros"))  # 3
```

```python
# --- Longest Increasing Subsequence, O(n log n) with binary search ---
from bisect import bisect_left
def length_of_lis(nums):
    tails = []                          # tails[k] = smallest tail of an LIS of length k+1
    for x in nums:
        i = bisect_left(tails, x)
        if i == len(tails):
            tails.append(x)
        else:
            tails[i] = x
    return len(tails)

print("LIS length:", length_of_lis([10, 9, 2, 5, 3, 7, 101, 18]))  # 4  ([2,3,7,101])

# --- Unique paths in a grid (2-D DP), O(m*n) ---
def unique_paths(m, n):
    dp = [1] * n
    for _ in range(1, m):
        for j in range(1, n):
            dp[j] += dp[j - 1]          # from top + from left
    return dp[-1]

print("unique paths 3x7:", unique_paths(3, 7))  # 28

# --- Top-down memoization demo with lru_cache ---
from functools import lru_cache
def count_ways_coins(coins, amount):
    @lru_cache(maxsize=None)
    def ways(i, remaining):
        if remaining == 0:
            return 1
        if i == len(coins) or remaining < 0:
            return 0
        return ways(i, remaining - coins[i]) + ways(i + 1, remaining)   # take / skip
    return ways(0, amount)

print("ways to make 5 from [1,2,5]:", count_ways_coins((1, 2, 5), 5))  # 4
```

## 🎯 Interview Tips — DP & Common Patterns
- **Start with recursion, then memoize.** If you can write the brute-force recursion and spot repeated subproblems, you have a DP.
- **State design is 80% of the battle.** Say out loud: *"`dp[i][j]` = the answer for prefix i of A and prefix j of B."*
- **Recognize the family:**
  - *Take / skip* an item → knapsack / subset sum.
  - *Match / delete / insert* on two strings → LCS / edit distance.
  - *At each step choose from k moves* → grid / stairs / jumps.
  - *Best ending here* → LIS, max subarray.
- **Space optimization:** if `dp[i]` uses only `dp[i-1]` (or the previous row), collapse to O(1) / O(n).
- **0/1 vs unbounded knapsack:** iterate capacity **downward** for 0/1 (each item once), **upward** for unbounded (reuse allowed).
- Verify **base cases** and **iteration order** — most DP bugs live there.
